```python
# ==========================================
# Load GEO Dataset
# ==========================================

geo = pd.read_csv("GEO.csv")

print("Dataset Loaded Successfully!")
```

```python
# ==========================================
# Display First 5 Rows
# ==========================================

geo.head()
```

```python
# ==========================================
# Dataset Shape
# ==========================================

rows, cols = geo.shape

print("Number of Rows :", rows)
print("Number of Columns :", cols)
```

```python
# ==========================================
# Column Names
# ==========================================

print("Column Names:\n")

for col in geo.columns:
    print(col)
```

```python
# ==========================================
# Dataset Information
# ==========================================

geo.info()
```

```python
# ==========================================
# Data Types
# ==========================================

geo.dtypes
```

```python
# ==========================================
# Statistical Summary
# ==========================================

geo.describe()
```

```python
# ==========================================
# Missing Values Analysis
# ==========================================

missing = geo.isnull().sum()

print("Missing Values Per Column:\n")
print(missing)

print("\nTotal Missing Values :", missing.sum())
```

```python
# ==========================================
# Duplicate Gene Analysis
# ==========================================

duplicate_genes = geo["Unnamed: 0"].duplicated().sum()

print("Duplicate Gene Names :", duplicate_genes)
```

```python
# ==========================================
# Unique Gene Count
# ==========================================

unique_genes = geo["Unnamed: 0"].nunique()

print("Unique Genes :", unique_genes)
```

```python
# ==========================================
# Expression Range
# ==========================================

minimum = geo.iloc[:,1:].min().min()
maximum = geo.iloc[:,1:].max().max()

print("Minimum Expression :", minimum)
print("Maximum Expression :", maximum)
```

```python
# ==========================================
# Memory Usage
# ==========================================

memory = geo.memory_usage(deep=True)

print(memory)

print("\nTotal Memory Used : {:.2f} MB".format(memory.sum()/1024**2))
```

```python
# ==========================================
# Dataset Orientation
# ==========================================

print("Current Dataset Orientation")

print("\nRows    -> Genes")
print("Columns -> Patient Samples")

print("\nMachine Learning requires:")

print("Rows    -> Samples")
print("Columns -> Features (Genes)")
```

```python
# ==========================================
# FINAL DATASET ANALYSIS REPORT
# ==========================================

print("="*60)
print("         GENETIC DATASET ANALYSIS REPORT")
print("="*60)

print(f"Dataset Name          : GEO")
print(f"Rows                  : {geo.shape[0]}")
print(f"Columns               : {geo.shape[1]}")
print(f"Genes                 : {unique_genes}")
print(f"Samples               : {geo.shape[1]-1}")
print(f"Missing Values        : {missing.sum()}")
print(f"Duplicate Genes       : {duplicate_genes}")
print(f"Minimum Expression    : {minimum}")
print(f"Maximum Expression    : {maximum}")

print("\nData Types")
print("--------------------------------")

print("Gene Names            : Object")
print("Expression Values     : Numeric")

print("\nDataset Orientation")
print("--------------------------------")

print("Rows                  : Genes")
print("Columns               : Samples")

print("\nConclusion")
print("--------------------------------")
print("The GEO dataset is a high-dimensional")
print("gene expression dataset suitable")
print("for Objective 1 preprocessing.")

print("="*60)
```

```python
# ==========================================
# Save Report to Text File
# ==========================================

report = f"""
============================================================
GENETIC DATASET ANALYSIS REPORT
============================================================

Dataset Name        : GEO

Rows                : {geo.shape[0]}
Columns             : {geo.shape[1]}
Genes               : {unique_genes}
Samples             : {geo.shape[1]-1}

Missing Values      : {missing.sum()}
Duplicate Genes     : {duplicate_genes}

Minimum Expression  : {minimum}
Maximum Expression  : {maximum}

Dataset Orientation

Rows    -> Genes
Columns -> Samples

Conclusion

The dataset is a high-dimensional gene expression
dataset suitable for preprocessing.
"""

with open("Genetic_Dataset_Report.txt", "w") as file:
    file.write(report)

print("Report Saved Successfully!")
```
